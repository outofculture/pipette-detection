```python

fig, ax = plt.subplots(3, len(datasets), figsize=(15, 6))

for i, ds in enumerate(datasets):
    idx, row, col = ds['position']
    row2, col2 = ds['rotated_position']
    ma = ds['data']
    ax[0, i].imshow(ma[idx].asarray())
    ax[0, i].scatter(col, row, c='r', s=40, marker='x')
    ax[1, i].imshow(ds['rotated_data'][idx, ..., 0])
    ax[1, i].scatter(col2, row2, c='r', s=40, marker='x')
    ax[2, i].plot(ds['z'])
    ax[2, i].scatter(idx, ds['z'][idx], c='b', s=40)


```
